```python
def get_PocketVec_pairs(protein1, protein2, PAIRS):

    # Sort pair
    protein1, protein2 = sorted([protein1, protein2])

    # Filter pairs
    PAIRS = PAIRS[(PAIRS['Protein1'] == protein1) & 
                  (PAIRS['Protein2'] == protein2)].sort_values(by='PocketVec distance')

    return PAIRS.reset_index(drop=True)


def get_PocketVec_triplets(protein1, protein2, protein3, TRIPLETS):

    # Sort triplet
    protein1, protein2, protein3 = sorted([protein1, protein2, protein3])

    # Filter triplet
    TRIPLETS = TRIPLETS[(TRIPLETS['protein1'] == protein1) & 
                        (TRIPLETS['protein2'] == protein2) &
                        (TRIPLETS['protein3'] == protein3)]
    
    return TRIPLETS.reset_index(drop=True)
```
