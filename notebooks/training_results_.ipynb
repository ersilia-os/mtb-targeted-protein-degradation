```python
import pandas as pd
import numpy as np
import tl2cgen
import os
```

```python
PATH_TO_MODELS = "/aloy/home/acomajuncosa/Ersilia/mtb/processed/unidock_docking/models"
BINS = ['bin_01', 'bin_05', "bin_1"]
aurocs = {}

for pocket in sorted(os.listdir(PATH_TO_MODELS))[:40]:

    if pocket != "alphafold2_P9WFS9_model_0_pocket_2":

        aurocs[pocket] = {}

        # For each bin
        for bin in BINS:

            r = open(os.path.join(PATH_TO_MODELS, pocket, bin, "AUROCs.csv")).readlines()[0]
            r = np.array(r.split(","), dtype=float)
            aurocs[pocket][bin] = r
```
