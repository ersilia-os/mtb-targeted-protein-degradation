```python
from rdkit import Chem
import random
import gzip

# Define paths
input_path = "../processed/enamine_characterization/conformations.sdf.gz"
output_path = "../processed/enamine_characterization/conformations_1000.sdf.gz"
N_SUBSAMPLE = 1000

# Read molecules
mols = []
with gzip.open(input_path, "rb") as f:
    suppl = Chem.ForwardSDMolSupplier(f, removeHs=False)
    for mol in suppl:
        if mol is not None:
            mols.append(mol)

print(f"Total molecules read: {len(mols)}")

# Sample molecules
random.seed(42)
subset = random.sample(mols, k=min(N_SUBSAMPLE, len(mols)))

# Write to new gzipped SDF
with gzip.open(output_path, "wt") as f_out:
    writer = Chem.SDWriter(f_out)
    for mol in subset:
        writer.write(mol)
    writer.close()

print(f"Saved {len(subset)} molecules to: {output_path}")
```
