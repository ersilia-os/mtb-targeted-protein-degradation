```python
from Bio.SeqRecord import SeqRecord
from Bio import SeqIO
import os
```

```python
### Merge all fasta files in a single one

fasta_files = [os.path.join("../data/sequences/fasta", i) for i in sorted(os.listdir(os.path.join("../data/sequences/fasta")))]
merged_fasta = os.path.join("../data/sequences", "merged.fasta")

with open(merged_fasta, "w") as out:
    for file in fasta_files:
        with open(os.path.join(file)) as f:
            out.write(f.read())
```

```python
### Create fasta files with N repetitions of the same sequence to run AF2 multimer

fasta_files = [os.path.join("../data/sequences/fasta", i) for i in sorted(os.listdir(os.path.join("../data/sequences/fasta")))]
repeated_fasta = os.path.join("../data/sequences/repeated_fasta")
N = 4

for file in fasta_files:

    filename = file.split("/")[-1]
    uni = filename.split(".")[0]
    record = next(SeqIO.parse(file, "fasta"))
    new_records = [SeqRecord(record.seq, id=f"{uni}--subunit{i+1}", description="") for i in range(4)]
    SeqIO.write(new_records, os.path.join(repeated_fasta, filename), "fasta")
```
