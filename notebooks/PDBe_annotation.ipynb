```python
import pandas as pd
import os
```

```python
# For how many proteins do we have a PDB structure?
pocket_detection_data = pd.read_csv("../processed/pocket_detection_data.csv")
proteins_pdb = sorted(os.listdir("../data/structures/pdbe_database"))

print(len(proteins_pdb))
print(len(pocket_detection_data[pocket_detection_data['Uniprot AC'].isin(proteins_pdb)] ))
```

```python
report = pd.read_csv('../processed/pdbe_annotation_report.csv', sep='\t')

# Removing single atoms
ignore = ['MG', 'CL', 'CA', 'K']
report = report[~report['HET RES'].isin(ignore)]

# Selecting only detected pockets at a distance < 6 A
report = report[report['Min Distance'] < 6]
```

```python
print(f"Number of proteins: {len(set(report['Uniprot ID']))}")
print(f"Number of pockets: {len(set(report['Pocket']))}")
```
