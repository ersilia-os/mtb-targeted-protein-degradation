```python
import os
import requests
from io import StringIO
from Bio.PDB.MMCIF2Dict import MMCIF2Dict
```

```python
def get_assembly_details_online(pdb_id):
    # Fetch the mmCIF content from RCSB
    url = f"https://files.rcsb.org/download/{pdb_id.upper()}.cif"
    response = requests.get(url)
    
    if response.status_code != 200:
        raise ValueError(f"Could not fetch CIF file for {pdb_id}")
    
    # Parse mmCIF directly from the content string
    mmcif_dict = MMCIF2Dict(StringIO(response.text))
    
    # Extract stoichiometry and oligomeric details
    stoichiometry = mmcif_dict.get('_pdbx_struct_assembly.stoichiometry')
    oligomeric_details = mmcif_dict.get('_pdbx_struct_assembly.oligomeric_details')
    
    return stoichiometry, oligomeric_details
```

```python
path = "../data/structures/pdbe_database/"
pdbs = [[uni,j] for uni in sorted(os.listdir(path)) for j in os.listdir(os.path.join(path, uni, uni + '_archive-PDB')) if '.txt' not in j]
```

```python
for pdb in pdbs:
    uni = pdb[0]
    if "ent" in pdb[1]:
        pdb = pdb[1].removesuffix('.ent').replace('pdb', "")
    elif 'cif' in pdb[1]:
        pdb = pdb[1].removesuffix('_updated.cif')
    else:
        print("PDB could not be parsed: " + pdb[1])
    stoich, oligomeric_details = get_assembly_details_online(pdb)
    print(uni, pdb, stoich, oligomeric_details)
```
