```python
PATH_TO_INFERENCE_OUTPUTS = os.path.join("/aloy/home/acomajuncosa/Ersilia/mtb/processed/unidock_docking/inferences_outputs")
print("Checking inference logs...")
for out in tqdm(sorted(os.listdir(PATH_TO_INFERENCE_OUTPUTS))):
    out = open(os.path.join(PATH_TO_INFERENCE_OUTPUTS, out), "r").readlines()
    out = [i for i in out if "warning" in i.lower() or "error" in i.lower()]
    if len(out) != 0:
        print(f"ERROR: CHECK FILE {out}")
print("If no message appeared, all is good")
```

```python
print("Reading inference results")
PATH_TO_INFERENCES = "/aloy/home/acomajuncosa/Ersilia/mtb/processed/unidock_docking/inferences"
PATH_TO_RESULTS = os.path.join("..", "processed", "unidock_docking", "inferences_top100k")
os.makedirs(PATH_TO_RESULTS, exist_ok=True)

# For each pocket
for pocket in sorted(os.listdir(PATH_TO_INFERENCES)):

    print(f"Processing {pocket} ...")

    # Load results
    inf = pd.read_csv(os.path.join(PATH_TO_INFERENCES, pocket))

    break

    # Select only the TOP-100k
    inf = inf.sort_values("probs", ascending=False)[:100000].reset_index(drop=True)

    # Save results
    inf.to_csv(os.path.join(PATH_TO_RESULTS, pocket.replace("bin_01", "bin_01_top100k").replace(".gz", "")), index=False)
```
